program ini di buat melakukan proses scraping terhadap website game digital 'steam' membuatnya menjadi data frame berdasarkan nama game, harga, taring, released, dan developer.
untuk selajutnya data tersebut akan di bersihkan dan di unggah ke sql dan akan di normalisasi kembali seluruh datanya. 


```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By 
from bs4 import BeautifulSoup
import time 

names = []
prices = []
ratings = []
released = []
publishers = []

#meminta driver untuk membuka web 
driver = webdriver.Chrome()
url = 'https://store.steampowered.com/category/action_fps'
driver.get(url)
time.sleep(30)

#melakukan scrolling untuk memuat data, karena sistem website steam tidak memiliki page sehingga perlu dilakukannya scrolling untuk memuat lebih banyak data 
for i in range(1, 15):
    jarak_scroll = i * 1000
    driver.execute_script(f"window.scrollTo(0, {jarak_scroll});")
    print(f"posisi {jarak_scroll}px")
    time.sleep(10)

driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
time.sleep(10)

soup = BeautifulSoup(driver.page_source, 'html.parser')

#melakukan boxes untuk mengambil data yang diinginkan, karena data yang diambil memiliki class yang sama sehingga perlu dilakukan box untuk mengambil data yang diinginkan
boxes = soup.find_all('div', {'class' : '_2hhNOdcC6yLwL_rugP3YLf PZY_7wH_NY7OgzH9MBiB8 LibraryAssetExpandedDisplay'})

#melakukan try except untuk mengambil data yang diinginkan, ini gunakan agar data data yang tidak memiliki isi tetap bisa diambil dan tidak error, sehingga data yang diambil tetap lengkap
for box in boxes:
    try:
        nama = box.find('img', {'class' : '_2eQ4mkpf4IzUp1e9NnM2Wr'}).get('alt')
        names.append(nama)
    except:
        names.append('None')    
    
    try:
        harga = box.find('div', {'class' : '_3j4dI1yA7cRfCvK8h406OB'}).get_text()
        prices.append(harga)
    except:
        prices.append('None')
    try:
        rating = box.find('div', {'class' : 'uEsfJ0VAuX37ItihZkK2J'}).get_text()
        ratings.append(rating)
    except:
        ratings.append('None')
    try:
        release = box.find('div', {'class' : 'vCEpeeiHJkcIDdtTkRfjT'}).get_text()
        released.append(release)
    except:
        released.append('None')
    try:
        publisher = box.find('div', {'class' : '_3tIbO7JWeYXTD8fDol5_-f'}).get_text()
        publishers.append(publisher)
    except:
        publishers.append('None')

time.sleep(10)
driver.quit()
```

```python
#pengecakan jumlah data yang di ambil untuk memastikan apakah sudah sesuai dengan kebutuhan jumlah data 
print(len(names))
print(len(prices))
print(len(ratings))
print(len(released))
print(len(publishers))

```

```python
#membuat data frame terhadap data yang sudah di scraping, agar data yang diambil bisa diolah lebih lanjut
df = pd.DataFrame({
    'Name' : names,
    'Price' : prices,
    'Rating' : ratings,
    'Release Date' : released,
    'Developer' : publishers
})

df
```

```python
#proses membersihkan data yang sudah di ambil tetapi bentuknya masih berantakan 

#menghilangkan kata "Release date: " agar lebih rapih dalam tabel
df['Release Date'] = df['Release Date'].str.replace('Release date: ', '', case=False)

#menghilangkan kata "Developed By:" agar lebih rapih dalam tabel
df['Developer'] = df['Developer'].str.replace('Developed By:', '', case=False)

df['Release Date'] = df['Release Date'].str.strip()
df['Developer'] = df['Developer'].str.strip()

#beberapa data di dalam release date perlru di modifikasi karna penataan tanggalnya berantakan di dalam tabel.
df['Release Date'] = df['Release Date'].str.replace('Release date: ', '', case=False)
df['Release Date'] = df['Release Date'].str.strip()
df['Release Date'] = df['Release Date'].str.replace('SEPT', 'Sep', case=False)
df['Release Date'] = pd.to_datetime(df['Release Date'], format='mixed', errors='coerce')


df
```
